```python
from rdkit.Chem import rdMolDescriptors, Descriptors

print(f"{'Name':<8} {'Formula':<15} {'MW':>8} {'LogP':>8} {'Atoms':>6} {'Rings':>6}")
print("-" * 50)
for name,smi in compounds.items():
    mol=Chem.MolFromSmiles(smi)

    if mol is None:
        print(f"{name:<8} ERROR- could not read")
        continue

    Formula=rdMolDescriptors.CalcMolFormula(mol)
    MW=Descriptors.MolWt(mol)
    LogP=Descriptors.MolLogP(mol)
    Atoms=mol.GetNumHeavyAtoms()
    Rings=rdMolDescriptors.CalcNumRings(mol)

    print(f"{name:<8} {Formula:<15} {MW:>8.2f} {LogP:>8.2f} {Atoms:>6} {Rings:>6}")
```

```python
from rdkit.Chem.inchi import MolToInchi, InchiToInchiKey

print(f"{'Name':<8} {'Canonical Smiles':<60} {'InchiKey'}")
print("-" * 50)

for name, smi in compounds.items():
    mol=Chem.MolFromSmiles(smi)
    if mol is None:
        print(f"{name:<8} ERROR")
        continue 

    Canonical=Chem.MolToSmiles(mol)
    inchi=MolToInchi(mol)
    key=InchiToInchiKey(inchi)

    print(f"{name:<8} {Canonical:<60} {key}")
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors, rdMolDescriptors
from rdkit.Chem.inchi import MolToInchi, InchiToInchiKey
import pandas as pd

rows = []

for name, smi in compounds.items():
    mol = Chem.MolFromSmiles(smi)

    if mol is None:
        print(f"Skipping {name} — invalid SMILES")
        continue

    row = {
        "Name"      : name,
        "SMILES"    : smi,
        "Canonical" : Chem.MolToSmiles(mol),
        "Formula"   : rdMolDescriptors.CalcMolFormula(mol),
        "MW"        : round(Descriptors.MolWt(mol), 2),
        "HeavyAtoms": mol.GetNumHeavyAtoms(),
        "Rings"     : rdMolDescriptors.CalcNumRings(mol),
        "InChIKey"  : InchiToInchiKey(MolToInchi(mol)),
    }
    rows.append(row)

df = pd.DataFrame(rows)

print(df.to_string(index=False))
```

```python
df.to_csv("EB_compounds_properties.csv",index=False)
print("Saved successfully!")
```
